**Q1.** What is Data Loading in Power BI and why is it considered the first step of analysis?

**Ans-** **Data Loading in Power BI** is the process of connecting to and importing data from various sources into Power BI so that the data can be prepared, analyzed, and visualized. Power BI supports many data sources, such as **Excel, CSV files, SQL databases, web sources, and cloud services**.

### **Why is data loading considered the first step of analysis?**

Data loading is considered the first step of data analysis because analysis cannot be performed without first making the required data available in Power BI. Once the data is loaded, it can be cleaned and transformed using Power Query, organized into a data model, analyzed using calculations and DAX, and finally presented through charts, reports, and dashboards.

The basic Power BI workflow is:

**Data Source → Data Loading → Data Cleaning & Transformation → Data Modeling → Analysis → Visualization → Report/Dashboard**

For example, if a company wants to analyze its sales performance, it may first load its sales data from an Excel file into Power BI. After loading the data, the analyst can remove errors, handle missing values, create relationships between tables, calculate KPIs, and create visualizations such as sales by region, monthly sales trends, and top-selling products.



**Q2.**  Explain the difference between “Load” and “Transform Data” in Power BI.

**Ans-** In Power BI, **Load** and **Transform Data** are two options used after connecting to a data source. They differ mainly in whether the data needs to be modified before being used for analysis.

| **Load**                                                         | **Transform Data**                                                                          |
| ---------------------------------------------------------------- | ------------------------------------------------------------------------------------------- |
| Loads the selected data directly into Power BI.                  | Opens the **Power Query Editor** to clean and modify the data before loading it.            |
| No major changes are made to the data before importing it.       | Allows you to perform various data cleaning and transformation operations.                  |
| Suitable when the data is already clean and properly structured. | Suitable when the data contains errors, missing values, duplicates, or formatting issues.   |
| Saves time when no transformation is required.                   | Provides greater control over the data preparation process.                                 |
| The data is loaded directly into the Power BI data model.        | After transformation, you click **Close & Apply** to load the prepared data into the model. |

### **Examples of Transform Data Operations**

Using **Transform Data**, we can:

* Remove duplicate rows
* Handle missing or null values
* Change data types
* Remove extra spaces using **Trim**
* Split or merge columns
* Rename columns
* Filter rows
* Replace incorrect values
* Create new columns
* Format and clean data

**Example :** Suppose an Excel file contains customer data with duplicate records, missing values, and incorrect data types.

- If the data is already clean, we can select **Load** and directly import it into Power BI.
- If the data needs cleaning, we should select **Transform Data**, make the necessary changes in Power Query Editor, and then select **Close & Apply** to load the cleaned data.




**Q3.** What is a Fact Table and a Dimension Table? Give examples from the dataset.

**Ans-** In Power BI, **a Fact Table** and **Dimension Table** are the two main types of tables used in a **Star Schema** for organizing data efficiently.

**1. Fact Table**

A **Fact Table** contains the **measurable and numerical information** related to business transactions or events. It usually contains **foreign keys** that connect it to different dimension tables.

For example, in a Sales dataset, the **Sales/Orders table** can be considered a Fact Table because it contains transaction-level information such as:

- Order ID
- Customer ID
- Product ID
- Order Date
- Quantity
- Sales
- Profit
- Discount

**Example:**

| Order ID | Customer ID | Product ID |	Quantity | Sales | Profit |
|----------|-------------|------------|----------|-------|--------|
|   1001	 |     C101    |	  P201	  |     2    |  5000 |	800   |
|   1002   |     C102    |   	P202    |   	1	   |  3000 |	500   |
|   1003   |   	 C101    |   	P203    |   	3	   |  4500 |	700   |

Here, **Quantity, Sales, and Profit** are facts or measurable values.

**2. Dimension Table**

A **Dimension Table** contains **descriptive information** that provides context to the data stored in the Fact Table. Dimension tables are mainly used to **filter, group, and categorize** the data during analysis.

For example, a **Product Dimension** may contain:

| Product ID | Product Name |	 Category  | Sub-Category |
|------------|--------------|------------|--------------|
|    P201    |  	Laptop    | Technology |	 Computers  |
|    P202    |  	Mobile	  | Technology |  	Phones    |
|    P203    |  	 Chair    | Furniture  |	  Chairs    |

**Q4.** Why is Star Schema preferred over Snowflake Schema in Power BI?

**Ans-** A Star Schema is preferred over a Snowflake Schema in Power BI because it provides **a simple, clear, and efficient structure for data analysis and reporting**. In a Star Schema, a central Fact Table is directly connected to multiple Dimension Tables.

The main reasons are:

**1. Easy to Read** – The structure is simple and organized, making the relationships between tables easy to see.

**2. Easy to Understand** – Users and analysts can easily understand the relationship between the Fact Table and Dimension Tables.

**3. Easy to Maintain** – Since there are fewer tables and relationships, making changes or managing the data model is easier.

**4. Better Performance** – Fewer relationships and joins can help improve query and report performance.

**5. Easy Analysis** – Dimension tables can be easily used for filtering, grouping, and slicing data in Power BI.
